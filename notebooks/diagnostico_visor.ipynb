Ejecuta las tres celdas **en orden** y anota cual se ve y cual no. Cada una prueba una
capa distinta, y el resultado dice donde esta el problema:

| se ve | no se ve | conclusion |
|---|---|---|
| 1, 2 y 3 | -- | el visor esta bien; el problema es del cuaderno 06 |
| 1 y 3 | **2** | ipywidgets y plotly van, pero **anywidget no**: es el fallo conocido del `FigureWidget` |
| 3 | 1 y 2 | **ipywidgets no carga**: falta la extension o el kernel no es el del venv |
| -- | 1, 2 y 3 | no hay salida de ningun tipo: revisa que el kernel sea `.venv` |


```python
# 1. ipywidgets puro -- tiene que verse un boton y una casilla
import ipywidgets as widgets
from IPython.display import display

display(widgets.VBox([widgets.Button(description='Si ves este boton, ipywidgets carga'),
                      widgets.Checkbox(value=True, description='y esta casilla')]))
```

```python
# 2. go.FigureWidget -- ESTA es la que usa el tablero del cuaderno 06.
# Desde plotly 6.0 la respalda `anywidget`, y es la que suele fallar en VS Code.
import plotly.graph_objects as go

fig = go.FigureWidget(data=[go.Scatter(x=[1, 2, 3], y=[2, 1, 3], mode='lines+markers')])
fig.update_layout(height=260, title='Si ves esta figura, anywidget carga')
display(fig)
```

```python
# 3. go.Figure normal (sin widget) -- no usa anywidget ni ipywidgets.
# Si esta se ve y la 2 no, el problema esta aislado en anywidget.
import plotly.graph_objects as go

go.Figure(data=[go.Bar(x=['a', 'b', 'c'], y=[3, 1, 2])]).update_layout(
    height=260, title='Si ves estas barras, plotly dibuja bien').show()
```
