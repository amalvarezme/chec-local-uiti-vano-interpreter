# 04 - Exploracion GEO para mapas de red CHEC

Este cuaderno analiza primero las capas geograficas en `data/GEO` antes de graficarlas. La idea es responder: que trae cada capa, que calidad espacial tiene, como se puede unir con `Indicadores_vano_v3.csv`, y que mapas conviene generar por circuito.

## 1. Setup

Las capas son shapefiles en WGS84 (`EPSG:4326`). Para calculos metricos se reproyectan temporalmente a un CRS estimado por `geopandas`.

```python
from __future__ import annotations

from pathlib import Path
import json
import warnings

import geopandas as gpd
import pandas as pd
import folium

pd.set_option("display.max_columns", 120)
warnings.filterwarnings("ignore", category=UserWarning)

# Sube desde el cwd hasta la raíz del repo (marcada por la carpeta src/).
ROOT = Path.cwd().resolve()
while not (ROOT / "src").is_dir() and ROOT.parent != ROOT:
    ROOT = ROOT.parent

GEO_DIR = ROOT / "data" / "GEO"
EVENTS_PATH = ROOT / "data" / "Indicadores_vano_v3.csv"
OUTPUT_DIR = ROOT / "reports" / "geo"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Shapefiles fuente: red de media tensión, transformadores e interruptores.
LINE_PATH = GEO_DIR / "MVLINSEC.shp"
TRANSFORMER_PATH = GEO_DIR / "GDBCHEC_TRANSFOR.shp"
SWITCH_PATH = GEO_DIR / "SWITCHES.shp"

print(f"GEO_DIR: {GEO_DIR}")
print(f"OUTPUT_DIR: {OUTPUT_DIR}")
```

## 2. Inventario de archivos

Antes de leer geometria conviene verificar que cada shapefile tenga sus archivos hermanos (`.shp`, `.shx`, `.dbf`, `.prj`, `.cpg`).

```python
# Inventario de tamaños/extensiones para detectar archivos faltantes o corruptos
# antes de invertir tiempo leyendo geometría con geopandas.
inventory = []
for path in sorted(GEO_DIR.glob("*")):
    inventory.append({
        "archivo": path.name,
        "extension": path.suffix.lower(),
        "tamano_mb": round(path.stat().st_size / 1024 / 1024, 3),
    })

df_inventory = pd.DataFrame(inventory)
display(df_inventory.sort_values(["archivo"]))

# Un shapefile depende de varios archivos hermanos; si falta alguno, geopandas
# puede fallar o leer con proyección/atributos incompletos.
shp_bases = sorted(path.stem for path in GEO_DIR.glob("*.shp"))
required_ext = {".shp", ".shx", ".dbf", ".prj", ".cpg"}
for base in shp_bases:
    present = {path.suffix.lower() for path in GEO_DIR.glob(f"{base}.*")}
    missing = sorted(required_ext - present)
    print(f"{base}: faltantes requeridos={missing or 'ninguno'}")
```

## 3. Carga de capas y perfil basico

- `MVLINSEC`: vanos/tramos de red de media tension como `LineString`.

El analisis inicial revisa registros, CRS, tipo geometrico, limites espaciales, nulos y columnas candidatas para unir los vanos con la base analitica.

```python
lineas = gpd.read_file(LINE_PATH)
trafos = gpd.read_file(TRANSFORMER_PATH)
switches = gpd.read_file(SWITCH_PATH)

# G3E_FID es el identificador nativo del SIG de CHEC (GE Smallworld); se renombra
# por capa para tener llaves explícitas al momento de cruzar/unir geometrías.
lineas["FID_VANO_GEO"] = lineas["G3E_FID"]
trafos["FID_TRAFO_GEO"] = trafos["G3E_FID"]
switches["FID_SWITCH_GEO"] = switches["G3E_FID"]

layers = {
    "vanos_mv": lineas,
    "transformadores": trafos,
    "interruptores": switches,
}

# Perfil rápido de calidad por capa: filas, CRS, tipos de geometría, geometrías
# vacías/nulas, límites espaciales y columnas con más nulos (candidatas a revisar).
def profile_layer(name: str, gdf: gpd.GeoDataFrame) -> dict:
    non_geom = gdf.drop(columns=gdf.geometry.name)
    return {
        "layer": name,
        "rows": int(len(gdf)),
        "columns": int(len(gdf.columns)),
        "crs": str(gdf.crs),
        "geometry_types": gdf.geometry.geom_type.value_counts(dropna=False).to_dict(),
        "empty_geometries": int(gdf.geometry.is_empty.sum()),
        "null_geometries": int(gdf.geometry.isna().sum()),
        "bounds": [round(float(x), 6) for x in gdf.total_bounds],
        "top_null_columns": non_geom.isna().mean().sort_values(ascending=False).head(12).round(3).to_dict(),
    }

profiles = pd.DataFrame([profile_layer(name, gdf) for name, gdf in layers.items()])
display(profiles)

for name, gdf in layers.items():
    print(f"\n=== {name} ===")
    print(list(gdf.columns))
    display(gdf.drop(columns="geometry").head(5))
```

## 4. Calidad espacial y medidas metricas

Las coordenadas estan en grados. Para medir longitudes se usa un CRS proyectado estimado. Esto permite comparar `LONGITUD`, `SHAPE_LEN` y longitud geometrica calculada.

```python
# Las coordenadas fuente están en grados (EPSG:4326); estimate_utm_crs elige un
# CRS métrico local para poder calcular longitudes reales en metros.
metric_crs = lineas.estimate_utm_crs() or "EPSG:3116"
print(f"CRS metrico estimado: {metric_crs}")

lineas_metricas = lineas.to_crs(metric_crs).copy()
lineas["longitud_geom_m"] = lineas_metricas.geometry.length

quality_rows = []
for name, gdf in layers.items():
    quality_rows.append({
        "layer": name,
        "geometrias_validas": int(gdf.geometry.is_valid.sum()),
        "geometrias_invalidas": int((~gdf.geometry.is_valid).sum()),
        "geometrias_vacias": int(gdf.geometry.is_empty.sum()),
        "geometrias_nulas": int(gdf.geometry.isna().sum()),
    })

display(pd.DataFrame(quality_rows))

cols_len = [col for col in ["CIRCUITO", "FID_VANO_GEO", "CODIGO", "LONGITUD", "SHAPE_LEN", "longitud_geom_m"] if col in lineas.columns]
display(lineas[cols_len].describe(include="all"))

# Compara la longitud declarada en el shapefile (LONGITUD) contra la longitud
# calculada de la geometría, para detectar tramos con datos de longitud inconsistentes.
if "LONGITUD" in lineas.columns:
    comp = lineas[["CIRCUITO", "FID_VANO_GEO", "LONGITUD", "longitud_geom_m"]].copy()
    comp["LONGITUD"] = pd.to_numeric(comp["LONGITUD"], errors="coerce")
    comp["delta_m"] = comp["LONGITUD"] - comp["longitud_geom_m"]
    display(comp[["LONGITUD", "longitud_geom_m", "delta_m"]].describe())
```

## 5. Resumen operativo por circuito

Este resumen ayuda a decidir que circuitos graficar primero: cantidad de vanos/tramos, longitud de red y municipios cubiertos.

```python
# Resumen por circuito: vanos/tramos, km de red y municipios que cubre.
lineas_summary = (
    lineas.assign(longitud_geom_km=lineas["longitud_geom_m"] / 1000)
    .groupby("CIRCUITO", dropna=False)
    .agg(
        n_vanos_geo=("FID_VANO_GEO", "count"),
        longitud_geom_km=("longitud_geom_km", "sum"),
        municipios=("MUNICIPIO", lambda s: sorted({str(x) for x in s.dropna().unique()})),
    )
    .reset_index()
)

# "ENERGIZADO" es un texto libre en el shapefile (no booleano); se detecta por
# substring case-insensitive en vez de comparar contra un valor exacto.
trafos_summary = (
    trafos.assign(CAPACIDAD_NUM=pd.to_numeric(trafos.get("CAPACIDAD_"), errors="coerce"))
    .groupby("CIRCUITO", dropna=False)
    .agg(
        n_transformadores_geo=("FID_TRAFO_GEO", "count"),
        capacidad_transformadores_kva=("CAPACIDAD_NUM", "sum"),
        transformadores_energizados=("ENERGIZADO", lambda s: int(s.astype(str).str.contains("ENERGIZADO", case=False, na=False).sum())),
    )
    .reset_index()
)

switches_summary = (
    switches.groupby("CIRCUITO", dropna=False)
    .agg(
        n_interruptores_geo=("FID_SWITCH_GEO", "count"),
        tipos_interruptor=("TIPO", lambda s: sorted({str(x) for x in s.dropna().unique()})),
        interruptores_energizados=("ENERGIZADO", lambda s: int(s.astype(str).str.contains("ENERGIZADO", case=False, na=False).sum())),
    )
    .reset_index()
)

geo_circuit_summary = (
    lineas_summary
    .merge(trafos_summary, on="CIRCUITO", how="left")
    .merge(switches_summary, on="CIRCUITO", how="left")
)
# Circuitos sin transformadores/interruptores generan NaN en el merge; se
# normalizan a 0 en vez de dejarlos como NaN para que el orden/top-N no falle.
for col in ["n_transformadores_geo", "capacidad_transformadores_kva", "transformadores_energizados", "n_interruptores_geo", "interruptores_energizados"]:
    geo_circuit_summary[col] = pd.to_numeric(geo_circuit_summary[col], errors="coerce").fillna(0)
geo_circuit_summary["tipos_interruptor"] = geo_circuit_summary["tipos_interruptor"].apply(lambda x: x if isinstance(x, list) else [])
geo_circuit_summary = geo_circuit_summary.sort_values("longitud_geom_km", ascending=False)
display(geo_circuit_summary.head(20))

# Artefacto persistido: insumo para decidir qué circuitos priorizar en mapas.
summary_path = OUTPUT_DIR / "geo_resumen_circuitos.csv"
geo_circuit_summary.to_csv(summary_path, index=False)
print(f"Resumen guardado: {summary_path}")
```

## 6. Llaves candidatas para unir con `Indicadores_vano_v3.csv`

Hipotesis iniciales:

- Eventos `FID_VANO` -> vanos/tramos `MVLINSEC.G3E_FID`.
- Eventos `CIRCUITO` -> `MVLINSEC.CIRCUITO`.

La celda siguiente cuantifica cobertura antes de construir mapas con eventos.

```python
# Normaliza IDs para poder comparar entre CSV (a veces con .0 flotante) y
# shapefile (string): recorta espacios, quita el sufijo ".0" y vacía valores nulos.
def norm_id(series: pd.Series) -> pd.Series:
    return (
        series.astype("string")
        .str.strip()
        .str.replace(r"\.0$", "", regex=True)
        .replace({"": pd.NA, "<NA>": pd.NA, "nan": pd.NA, "None": pd.NA})
    )

event_cols = ["CIRCUITO", "FID_VANO", "UITI_VANO", "FECHA"]
events = pd.read_csv(EVENTS_PATH, usecols=lambda c: c in event_cols)
for col in ["FID_VANO", "CIRCUITO"]:
    if col in events.columns:
        events[col] = norm_id(events[col])

line_keys = lineas[["G3E_FID", "CIRCUITO"]].copy()
for col in line_keys.columns:
    line_keys[col] = norm_id(line_keys[col])

# Cuantifica qué tan bien casan los eventos (CSV analítico) contra la geometría
# (shapefile) por dos posibles llaves, para decidir a qué nivel de detalle
# conviene mapear (por vano individual o solo por circuito).
coverage = []
tests = [
    ("FID_VANO -> MVLINSEC.G3E_FID", events.get("FID_VANO"), line_keys["G3E_FID"]),
    ("CIRCUITO -> lineas.CIRCUITO", events.get("CIRCUITO"), line_keys["CIRCUITO"]),
]
for label, left, right in tests:
    if left is None:
        continue
    left_unique = set(left.dropna().astype(str).unique())
    right_unique = set(right.dropna().astype(str).unique())
    matched = left_unique & right_unique
    coverage.append({
        "union": label,
        "valores_eventos": len(left_unique),
        "valores_geo": len(right_unique),
        "valores_eventos_con_match": len(matched),
        "cobertura_eventos_unique": round(len(matched) / max(len(left_unique), 1), 4),
    })

display(pd.DataFrame(coverage).sort_values("cobertura_eventos_unique", ascending=False))
```

## 7. Agregacion de eventos sobre geometria

Si la cobertura de `FID_VANO -> lineas.G3E_FID` es buena, se puede pintar cada tramo por eventos, suma de `UITI_VANO` o promedio. Si no, se puede empezar por mapas a nivel circuito.

```python
# Agrega eventos UITI_VANO por FID_VANO para poder colorear cada tramo
# geométrico por su severidad acumulada/promedio.
events_agg_vano = (
    events.assign(
        FID_VANO=norm_id(events["FID_VANO"]),
        UITI_VANO_NUM=pd.to_numeric(events.get("UITI_VANO"), errors="coerce"),
    )
    .groupby("FID_VANO", dropna=True)
    .agg(
        n_eventos=("FID_VANO", "size"),
        uiti_vano_total=("UITI_VANO_NUM", "sum"),
        uiti_vano_prom=("UITI_VANO_NUM", "mean"),
        circuitos_eventos=("CIRCUITO", lambda s: sorted({str(x) for x in s.dropna().unique()})),
    )
    .reset_index()
)

lineas_eventos = lineas.copy()
lineas_eventos["G3E_FID_NORM"] = norm_id(lineas_eventos["G3E_FID"])
lineas_eventos["FID_VANO_GEO"] = lineas_eventos["G3E_FID_NORM"]
lineas_eventos = lineas_eventos.merge(events_agg_vano, left_on="G3E_FID_NORM", right_on="FID_VANO", how="left")
# Tramos sin eventos asociados quedan en 0, no en NaN, para que ordenar/graficar no falle.
lineas_eventos["n_eventos"] = lineas_eventos["n_eventos"].fillna(0).astype(int)
lineas_eventos["uiti_vano_total"] = lineas_eventos["uiti_vano_total"].fillna(0.0)

display(lineas_eventos[["CIRCUITO", "FID_VANO_GEO", "CODIGO", "n_eventos", "uiti_vano_total", "uiti_vano_prom"]].sort_values("uiti_vano_total", ascending=False).head(20))
```

## 8. Mapas exploratorios

Primer mapa recomendado: circuito seleccionado con vanos/tramos coloreados por `uiti_vano_total`. Esto permite validar visualmente si la geometria corresponde a los vanos/eventos esperados antes de diseñar mapas definitivos.

```python
# Circuito de ejemplo elegido para validar visualmente el mapa antes de
# generalizar a otros circuitos.
CIRCUITO_MAPA = "DON23L13"

# Rojo/grueso para tramos con al menos un evento UITI_VANO acumulado, azul/fino
# para tramos sin eventos registrados.
def style_line(feature):
    value = feature["properties"].get("uiti_vano_total") or 0
    if value > 0:
        color = "#dc2626"
        weight = 4
    else:
        color = "#2563eb"
        weight = 2
    return {"color": color, "weight": weight, "opacity": 0.75}

def _available_columns(gdf: gpd.GeoDataFrame, columns: list[str]) -> list[str]:
    return [col for col in columns if col in gdf.columns]

def _safe_text(value) -> str:
    if pd.isna(value):
        return ""
    return str(value)

def _popup_html(row, fields: list[tuple[str, str]], title: str) -> str:
    items = []
    for col, label in fields:
        value = row.get(col, "")
        text = _safe_text(value)
        if text:
            items.append(f"<tr><th style='text-align:left;padding-right:8px'>{label}</th><td>{text}</td></tr>")
    return f"<strong>{title}</strong><table>{''.join(items)}</table>"

# Capa de puntos genérica (usada para transformadores e interruptores): cada
# fila se reproyecta a EPSG:4326 (folium solo entiende lat/lon) y se dibuja
# como CircleMarker con popup de atributos.
def _add_point_layer(
    fmap: folium.Map,
    gdf: gpd.GeoDataFrame,
    *,
    name: str,
    color: str,
    radius: int,
    popup_fields: list[tuple[str, str]],
) -> None:
    if gdf.empty:
        return
    points = gdf.to_crs("EPSG:4326") if str(gdf.crs) != "EPSG:4326" else gdf
    group = folium.FeatureGroup(name=name, show=True)
    for _, row in points.iterrows():
        geom = row.geometry
        if geom is None or geom.is_empty:
            continue
        folium.CircleMarker(
            location=[geom.y, geom.x],
            radius=radius,
            color=color,
            fill=True,
            fill_color=color,
            fill_opacity=0.85,
            weight=1,
            tooltip=f"{name}: {_safe_text(row.get('CODIGO', row.get('G3E_FID', '')))}",
            popup=folium.Popup(_popup_html(row, popup_fields, name), max_width=420),
        ).add_to(group)
    group.add_to(fmap)

# Construye un mapa folium con las 3 capas (vanos, transformadores,
# interruptores) filtradas a un solo circuito, centrado en su bounding box.
def make_circuit_map(circuito: str, max_features: int | None = None) -> folium.Map:
    g_lines = lineas_eventos[lineas_eventos["CIRCUITO"].astype(str).eq(str(circuito))].copy()
    g_trafos = trafos[trafos["CIRCUITO"].astype(str).eq(str(circuito))].copy()
    g_switches = switches[switches["CIRCUITO"].astype(str).eq(str(circuito))].copy()
    if max_features is not None:
        g_lines = g_lines.head(max_features)
        g_trafos = g_trafos.head(max_features)
        g_switches = g_switches.head(max_features)
    if g_lines.empty and g_trafos.empty and g_switches.empty:
        raise ValueError(f"No hay geometria para circuito {circuito}")

    bounds_source = pd.concat([
        gdf[["geometry"]].to_crs("EPSG:4326") if str(gdf.crs) != "EPSG:4326" else gdf[["geometry"]]
        for gdf in [g_lines, g_trafos, g_switches]
        if not gdf.empty
    ], ignore_index=True)
    bounds = gpd.GeoDataFrame(bounds_source, geometry="geometry", crs="EPSG:4326").total_bounds
    center = [(bounds[1] + bounds[3]) / 2, (bounds[0] + bounds[2]) / 2]
    fmap = folium.Map(location=center, zoom_start=12, tiles="CartoDB positron")

    if not g_lines.empty:
        folium.GeoJson(
            g_lines[_available_columns(g_lines, ["FID_VANO_GEO", "CODIGO", "CIRCUITO", "n_eventos", "uiti_vano_total", "geometry"])],
            name="Vanos / tramos MV",
            style_function=style_line,
            tooltip=folium.GeoJsonTooltip(fields=_available_columns(g_lines, ["FID_VANO_GEO", "CODIGO", "CIRCUITO", "n_eventos", "uiti_vano_total"])),
        ).add_to(fmap)

    _add_point_layer(
        fmap,
        g_trafos,
        name="Transformadores",
        color="#f59e0b",
        radius=5,
        popup_fields=[
            ("FID_TRAFO_GEO", "FID trafo"),
            ("CODIGO", "Código"),
            ("CIRCUITO", "Circuito"),
            ("CAPACIDAD_", "Capacidad"),
            ("FASES", "Fases"),
            ("MUNICIPIO", "Municipio"),
            ("DIRECCION", "Dirección"),
            ("ENERGIZADO", "Energizado"),
            ("EST_OPERAT", "Estado operativo"),
        ],
    )
    _add_point_layer(
        fmap,
        g_switches,
        name="Interruptores / switches",
        color="#7c3aed",
        radius=4,
        popup_fields=[
            ("FID_SWITCH_GEO", "FID switch"),
            ("CODIGO", "Código"),
            ("TIPO", "Tipo"),
            ("ELEMENTO", "Elemento"),
            ("CIRCUITO", "Circuito"),
            ("CAPACIDAD_", "Capacidad"),
            ("FASES", "Fases"),
            ("MUNICIPIO", "Municipio"),
            ("DIRECCION", "Dirección"),
            ("ENERGIZADO", "Energizado"),
            ("EST_OPERAT", "Estado operativo"),
        ],
    )

    folium.LayerControl(collapsed=False).add_to(fmap)
    fmap.fit_bounds([[bounds[1], bounds[0]], [bounds[3], bounds[2]]])
    return fmap

mapa = make_circuit_map(CIRCUITO_MAPA)
map_path = OUTPUT_DIR / f"mapa_geo_{CIRCUITO_MAPA}.html"
mapa.save(map_path)
print(f"Mapa guardado: {map_path}")
print(
    f"Capas para {CIRCUITO_MAPA}: "
    f"vanos={len(lineas_eventos[lineas_eventos['CIRCUITO'].astype(str).eq(str(CIRCUITO_MAPA))])}, "
    f"trafos={len(trafos[trafos['CIRCUITO'].astype(str).eq(str(CIRCUITO_MAPA))])}, "
    f"interruptores={len(switches[switches['CIRCUITO'].astype(str).eq(str(CIRCUITO_MAPA))])}"
)
mapa
```

## 9. Como decidir que graficar despues

Use esta lista antes de construir mapas finales:

1. **Cobertura de llaves:** confirmar si `FID_VANO -> G3E_FID` cubre la mayoria de vanos con eventos. Si no, usar `CIRCUITO` como primer nivel.
2. **Calidad geometrica:** revisar geometrias vacias/invalidas y longitudes atipicas.
3. **Escala del mapa:** para red completa usar agregacion por circuito; para diagnostico usar circuito individual.
4. **Variable visual:** empezar con `n_eventos`, `uiti_vano_total` y `uiti_vano_prom` sobre vanos/tramos.
5. **Comparacion con cuadernos existentes:** los mapas deben complementar, no reemplazar, los graficos de puntos criticos e inferencia.

Mapas candidatos:

- Red por circuito con color por `UITI_VANO` total.
- Circuito seleccionado con vanos/tramos con eventos resaltados.
- Mapa de calidad de datos: tramos sin evento asociado, sin circuito o con longitud inconsistente.

## 10. Mapa climático temporal por circuito

Este mapa conserva las capas geográficas del circuito y colorea sus tramos según una variable climática. El control temporal recorre únicamente las fechas disponibles; no representa interpolación espacial continua.

```python
from branca.colormap import LinearColormap
import ipywidgets as widgets

TEMPORAL_CIRCUIT = CIRCUITO_MAPA
TEMPORAL_EXPORT_HTML = True
# Periodo de reproducción ISO 8601 (1 hora), coherente con la resolución horaria
# de las columnas climáticas <variable>_<lag_hora>.
TEMPORAL_PLAYBACK_PERIOD = "PT1H"

# Etiqueta en español -> prefijo real de columna en Indicadores_vano_v3.csv.
CLIMATE_VARIABLE_OPTIONS = {
    "Temperatura": "temp",
    "Precipitación": "prep",
    "Presión": "pres",
    "Presión superficial": "sp",
    "Humedad relativa": "rh",
    "Radiación solar": "solar_rad",
    "Ráfaga de viento": "wind_gust_spd",
    "Velocidad del viento": "wind_spd",
    "Nubosidad": "clouds",
}
climate_variable_selector = widgets.Dropdown(
    options=CLIMATE_VARIABLE_OPTIONS,
    value="temp",
    description="Variable:",
    style={"description_width": "initial"},
)
display(climate_variable_selector)
print("Seleccioná la variable y luego ejecutá las dos celdas siguientes.")
```

```python
TEMPORAL_CLIMATE_VARIABLE = climate_variable_selector.value
# Cada variable climática tiene 25 columnas de lag horario (0 a 24h antes del evento).
climate_columns = [f"{TEMPORAL_CLIMATE_VARIABLE}_{lag}" for lag in range(25)]
available_columns = pd.read_csv(EVENTS_PATH, nrows=0).columns
required_columns = {"CIRCUITO", "FID_VANO", "FECHA", *climate_columns}
missing_columns = sorted(required_columns - set(available_columns))
if missing_columns:
    raise ValueError(f"Columnas no disponibles: {missing_columns}")

climate = pd.read_csv(EVENTS_PATH, usecols=sorted(required_columns))
climate["CIRCUITO"] = norm_id(climate["CIRCUITO"])
climate["FID_VANO_NORM"] = norm_id(climate["FID_VANO"])
climate["FECHA"] = pd.to_datetime(climate["FECHA"], errors="coerce")

selected_climate = climate.loc[
    climate["CIRCUITO"].eq(str(TEMPORAL_CIRCUIT))
    & climate["FECHA"].notna()
    & climate["FID_VANO_NORM"].notna()
].copy()
if selected_climate.empty:
    raise ValueError(f"No hay eventos válidos para el circuito {TEMPORAL_CIRCUIT}")

# "Despliega" el formato ancho (una columna por lag horario) a formato largo
# para poder reconstruir una serie de tiempo real por vano.
hourly_climate = selected_climate.melt(
    id_vars=["FID_VANO_NORM", "FECHA"],
    value_vars=climate_columns,
    var_name="CLIMATE_COLUMN",
    value_name="CLIMATE_VALUE",
)
hourly_climate["LAG_HOURS"] = (
    hourly_climate["CLIMATE_COLUMN"].str.rsplit("_", n=1).str[-1].astype(int)
)
hourly_climate["CLIMATE_VALUE"] = pd.to_numeric(
    hourly_climate["CLIMATE_VALUE"], errors="coerce"
)
# Cada observación climática está anclada a FECHA (el evento) menos su lag,
# lo que reconstruye el timestamp real al que corresponde ese valor horario.
hourly_climate["TIMESTAMP"] = (
    hourly_climate["FECHA"]
    - pd.to_timedelta(hourly_climate["LAG_HOURS"], unit="h")
).dt.floor("h")
hourly_climate = hourly_climate.dropna(
    subset=["TIMESTAMP", "FID_VANO_NORM", "CLIMATE_VALUE"]
)

# Si varios eventos aportan el mismo vano y hora, se usa su promedio horario.
hourly_climate = (
    hourly_climate
    .groupby(["TIMESTAMP", "FID_VANO_NORM"], as_index=False)["CLIMATE_VALUE"]
    .mean()
    .sort_values(["TIMESTAMP", "FID_VANO_NORM"])
)
if hourly_climate.empty:
    raise ValueError(
        f"No hay valores horarios válidos para {TEMPORAL_CIRCUIT} / "
        f"{TEMPORAL_CLIMATE_VARIABLE}_0…{TEMPORAL_CLIMATE_VARIABLE}_24"
    )

network = lineas.loc[
    lineas["CIRCUITO"].astype(str).eq(str(TEMPORAL_CIRCUIT))
].copy()
network["FID_VANO_NORM"] = norm_id(network["G3E_FID"])
network = network.to_crs("EPSG:4326")
if network.empty:
    raise ValueError(f"No hay geometría para el circuito {TEMPORAL_CIRCUIT}")

temporal_network = network[["FID_VANO_NORM", "geometry"]].merge(
    hourly_climate, on="FID_VANO_NORM", how="inner"
)
if temporal_network.empty:
    raise ValueError("No fue posible unir los valores climáticos con los tramos geográficos")

color_min = float(temporal_network["CLIMATE_VALUE"].min())
color_max = float(temporal_network["CLIMATE_VALUE"].max())
if color_min == color_max:
    # Evita un rango degenerado (min==max) que rompería LinearColormap.
    color_min -= 0.5
    color_max += 0.5
color_values = [
    "#313695", "#4575b4", "#74add1", "#abd9e9", "#fee090",
    "#fdae61", "#f46d43", "#d73027", "#a50026",
]
color_scale = LinearColormap(color_values, vmin=color_min, vmax=color_max)

available_hours = [
    value.strftime("%Y-%m-%d %H:%M")
    for value in temporal_network["TIMESTAMP"].drop_duplicates().sort_values()
]
print(
    f"Circuito={TEMPORAL_CIRCUIT} | variable={TEMPORAL_CLIMATE_VARIABLE} | "
    f"horas={len(available_hours)} | tramos con datos={temporal_network['FID_VANO_NORM'].nunique()}"
)

# Payload compacto: una geometría por vano y solamente horas con observaciones.
# Se separa la geometría (fija) de los valores por hora, para no repetir la
# geometría en cada frame del control temporal del mapa (celda siguiente).
network_for_map = (
    network[["FID_VANO_NORM", "geometry"]]
    .dropna(subset=["FID_VANO_NORM", "geometry"])
    .drop_duplicates(subset=["FID_VANO_NORM"])
)
network_geojson = json.loads(network_for_map.to_json())
values_by_hour = {}
for timestamp, frame in temporal_network.groupby("TIMESTAMP", sort=True):
    values_by_hour[timestamp.strftime("%Y-%m-%d %H:%M")] = {
        str(row.FID_VANO_NORM): {
            "value": round(float(row.CLIMATE_VALUE), 6),
            "color": color_scale(float(row.CLIMATE_VALUE)),
        }
        for row in frame.itertuples(index=False)
    }
```

```python
# Reutiliza make_circuit_map solo para heredar el mismo centrado/zoom del mapa
# estático, pero descarta sus capas base/GeoJson porque este mapa las reemplaza
# por una versión controlada por JS con recorrido temporal.
temporal_climate_map = make_circuit_map(TEMPORAL_CIRCUIT)

for child_key, child in list(temporal_climate_map._children.items()):
    if isinstance(child, (folium.raster_layers.TileLayer, folium.map.LayerControl, folium.features.GeoJson)):
        temporal_climate_map._children.pop(child_key)
folium.TileLayer("OpenStreetMap", name="OpenStreetMap", control=False).add_to(temporal_climate_map)
folium.GeoJson(
    network_for_map,
    name="Red base",
    style_function=lambda _feature: {"color": "#6b7280", "weight": 2, "opacity": 0.30},
).add_to(temporal_climate_map)

# El slider/play temporal se implementa en JS puro (Leaflet, ya cargado por
# folium) porque ipywidgets no persiste su estado en el HTML exportado a disco;
# geometry/valuesByHour se embeben como JSON literal para que el HTML standalone
# funcione sin servidor ni kernel de Python detrás.
map_name = temporal_climate_map.get_name()
geometry_json = json.dumps(network_geojson, ensure_ascii=False, separators=(",", ":"))
values_json = json.dumps(values_by_hour, ensure_ascii=False, separators=(",", ":"))
variable_json = json.dumps(TEMPORAL_CLIMATE_VARIABLE)
# El control recorre únicamente horas con datos; no genera horas vacías entre eventos.
temporal_script = f"""
document.addEventListener("DOMContentLoaded", function() {{
  const map = {map_name};
  const geometry = {geometry_json};
  const valuesByHour = {values_json};
  const variable = {variable_json};
  const hours = Object.keys(valuesByHour).sort();
  let climateLayer = null;
  let timer = null;
  let playbackDelay = 900;

  const control = L.control({{position: "bottomleft"}});
  control.onAdd = function() {{
    const box = L.DomUtil.create("div", "climate-time-control");
    box.style.cssText = "background:rgba(255,255,255,.96);padding:10px 12px;border-radius:6px;"
      + "box-shadow:0 1px 5px rgba(0,0,0,.35);min-width:330px;font:12px sans-serif;";
    box.innerHTML =
      '<div style="font-weight:600;margin-bottom:6px">Evolución horaria — ' + variable + '</div>'
      + '<button type="button" data-role="play" style="width:62px;margin-right:8px">▶ Play</button>'
      + '<input data-role="slider" type="range" min="0" max="' + Math.max(hours.length - 1, 0)
      + '" value="0" step="1" style="width:220px;vertical-align:middle">'
      + '<label style="margin-left:10px">Velocidad ' +
        '<select data-role="speed">' +
        '<option value="1800">0.5×</option>' +
        '<option value="900" selected>1×</option>' +
        '<option value="450">2×</option>' +
        '<option value="225">4×</option>' +
        '</select></label>'
      + '<div data-role="label" style="margin-top:6px;font-weight:600"></div>';
    L.DomEvent.disableClickPropagation(box);
    L.DomEvent.disableScrollPropagation(box);
    return box;
  }};
  control.addTo(map);

  const container = control.getContainer();
  const slider = container.querySelector('[data-role="slider"]');
  const label = container.querySelector('[data-role="label"]');
  const play = container.querySelector('[data-role="play"]');
  const speed = container.querySelector('[data-role="speed"]');

  function render(index) {{
    if (!hours.length) return;
    const hour = hours[index];
    const values = valuesByHour[hour] || {{}};
    if (climateLayer) map.removeLayer(climateLayer);
    climateLayer = L.geoJSON(geometry, {{
      style: function(feature) {{
        const fid = String(feature.properties.FID_VANO_NORM);
        const item = values[fid];
        return item
          ? {{color:item.color,weight:9,opacity:1.0}}
          : {{color:"#9ca3af",weight:2,opacity:0.14}};
      }},
      onEachFeature: function(feature, layer) {{
        const fid = String(feature.properties.FID_VANO_NORM);
        const item = values[fid];
        if (item) {{
          layer.bindPopup(
            "<strong>Mapa climático por tramo</strong><br>"
            + "Hora: " + hour + "<br>FID_VANO: " + fid + "<br>"
            + variable + ": " + Number(item.value).toFixed(3)
          );
        }}
      }}
    }}).addTo(map);
    climateLayer.bringToFront();
    label.textContent = hour + " · " + Object.keys(values).length + " vanos con datos";
  }}

  function startPlayback() {{
    timer = setInterval(function() {{
      slider.value = (Number(slider.value) + 1) % hours.length;
      render(Number(slider.value));
    }}, playbackDelay);
  }}

  slider.addEventListener("input", function() {{ render(Number(slider.value)); }});
  speed.addEventListener("change", function() {{
    playbackDelay = Number(speed.value);
    if (timer) {{ clearInterval(timer); startPlayback(); }}
  }});
  play.addEventListener("click", function() {{
    if (timer) {{
      clearInterval(timer); timer = null; play.textContent = "▶ Play"; return;
    }}
    play.textContent = "⏸ Pausa";
    startPlayback();
  }});
  render(0);
}});
"""
temporal_climate_map.get_root().script.add_child(folium.Element(temporal_script))

vertical_gradient = ", ".join(reversed(color_values))
legend_html = f"""
<div style="position:fixed;top:110px;right:18px;z-index:9999;background:rgba(255,255,255,.94);
            padding:10px;border-radius:6px;box-shadow:0 1px 5px rgba(0,0,0,.35);font-size:12px;">
  <div style="font-weight:600;margin-bottom:6px;">{TEMPORAL_CLIMATE_VARIABLE}</div>
  <div style="display:flex;gap:6px;align-items:stretch;height:190px;">
    <div style="display:flex;flex-direction:column;justify-content:space-between;">
      <span>{color_max:.2f}</span><span>{color_min:.2f}</span>
    </div>
    <div style="width:18px;border:1px solid #666;background:linear-gradient(to top,{vertical_gradient});"></div>
  </div>
</div>
"""
temporal_climate_map.get_root().html.add_child(folium.Element(legend_html))

temporal_map_path = OUTPUT_DIR / (
    f"mapa_clima_temporal_{TEMPORAL_CIRCUIT}_{TEMPORAL_CLIMATE_VARIABLE}.html"
)
if TEMPORAL_EXPORT_HTML:
    temporal_climate_map.save(temporal_map_path)
    print(f"Mapa temporal guardado: {temporal_map_path}")

display(temporal_climate_map)
```
